# 3.- Definición inicial de criterio de salida para el (o los) algoritmos de Machine Learning a utilizar.

### a.- Selección inicial de algoritmo (máximo 3).

- **Regresión Lineal:** Permitirá establecer un modelo base y analizar si existe una relación lineal entre las características musicales y la popularidad.

- **Árbol de Decisión para Regresión:** Permite capturar relaciones no lineales y generar predicciones mediante reglas basadas en las características de las canciones.

- **Random Forest Regressor:** Combina múltiples árboles de decisión, permitiendo modelar relaciones más complejas y reducir la dependencia de un único árbol.

### b.- Determinación y justificación de las métricas seleccionadas para evaluar el rendimiento del modelo.

- **MAE (Error Absoluto Medio):** Calcula el promedio de la diferencia absoluta entre los valores reales y los valores predichos.

- **RMSE (Raíz del Error Cuadrático Medio):** Calcula el error de las predicciones dando mayor importancia a los errores grandes.

- **$R^2$ (Coeficiente de determinación):** Indica qué proporción de la variabilidad de "popularity" logra explicar el modelo.

### ¿Por qué se eligieron?

MAE nos permite conocer aproximadamente cuántos puntos de "popularity" se equivoca el modelo en promedio. RMSE nos permite identificar si el modelo está realizando algunas predicciones considerablemente alejadas de la popularidad real. Y $R^2$ nos permite evaluar qué tan bien las características utilizadas explican las diferencias de popularidad entre las canciones.

### c.- Criterio de Salida: Umbral de métricas seleccionadas para considerar el modelo desarrollado como válido.

Para considerar un modelo como válido, se evaluará su desempeño utilizando las métricas MAE, RMSE y R², seleccionadas previamente.

Se considerará como mejor modelo aquel que presente:

- Un MAE menor, indicando un menor error promedio en los puntos de popularidad.
- Un RMSE menor, indicando una menor presencia de errores grandes.
- Un R² mayor, indicando una mayor capacidad para explicar la variabilidad de la variable popularity.

Como criterio de salida, el modelo seleccionado deberá superar el desempeño de los demás modelos evaluados en la mayor parte de estas métricas y presentar resultados consistentes sobre el conjunto de prueba.

En este proyecto, la Regresión Lineal obtuvo el mejor desempeño entre los modelos evaluados, con un MAE de 12,01, un RMSE de 16,85 y un R² de 0,3209, superando al Árbol de Decisión y al Random Forest en las tres métricas utilizadas. Por lo tanto, se selecciona como el modelo que mejor cumple con el criterio de salida establecido.

## Detección de valores atípicos:

 Se utilizaron diagramas de caja (boxplots) y el método del rango intercuartílico (IQR) para detectar posibles valores atípicos en las características musicales del dataset. Se consideraron como posibles valores atípicos aquellos registros ubicados por debajo de Q1 - 1.5 × IQR o por encima de Q3 + 1.5 × IQR.

```python
caract = ["danceability", "energy", "loudness", "speechiness", "acousticness", "instrumentalness", "liveness", "valence", "tempo"]
for columna in caract:

    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = df[
        (df[columna] < limite_inferior) |
        (df[columna] > limite_superior)
    ]

    print(f"{columna}: {len(outliers)} valores atípicos\n")
```

# 6.- Preparación inicial de los Datos.

## Limpieza de datos


```python
# Eliminación de la columna Unnamed
df_clean = df.drop(columns=['Unnamed: 0'])

# Tratamiento de valores nulos
df_clean = df_clean.dropna(subset=['artists', 'album_name', 'track_name'])

# Tratamiento de duplicados
df_clean = df_clean.drop_duplicates(subset=['track_id'])

print(f'Filas antes de la limpieza: {df.shape[0]}')
print(f'Filas dsp de la limpieza: {df_clean.shape[0]}')
print(f'Filas eliminadas: {df.shape[0] - df_clean.shape[0]}')

print('-.'*20)

print(f'Columnas antes de la limpieza: {df.shape[1]}')
print(f'Columnas dsp de la limpieza: {df_clean.shape[1]}')
print(f'Columnas eliminadas: {df.shape[1] - df_clean.shape[1]}')
df_clean.head()
```

## Seleccion de variables y codificación de categorias

```python
# Transformación de la variable 'explicit'
df_clean['explicit'] = df_clean['explicit'].astype(int)

# Aplicación de One-Hot Encoding a la variable de género musical
df_encoded = pd.get_dummies(df_clean, columns=['track_genre'], drop_first=True)


features_to_drop = ['track_id', 'artists', 'album_name', 'track_name', 'popularity']

X = df_encoded.drop(columns=features_to_drop)
y = df_encoded['popularity']

df_encoded.head()
```

## División del dataset

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)

print(f"Muestras de entrenamiento (X_train): {X_train.shape[0]}")
print(f"Muestras de prueba (X_test): {X_test.shape[0]}")
```

## Escalamiento y normalización


```python
from sklearn.preprocessing import StandardScaler

# Lista de columnas num continuas
num_cols = [
    'duration_ms', 'danceability', 'energy', 'key', 'loudness',
    'mode', 'speechiness', 'acousticness', 'instrumentalness',
    'liveness', 'valence', 'tempo', 'time_signature']

scaler = StandardScaler()

# Copia de los conjuntos para usar los nombres de columnas
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

# Ajuste y transformación en Train
X_train_scaled[num_cols] = scaler.fit_transform(X_train[num_cols])
X_test_scaled[num_cols] = scaler.transform(X_test[num_cols])

X_train_scaled.head()
```

# 7.- Modelamiento y Evaluación de los Modelos.

##Modelo 1: Regresión Lineal

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

# Instanciación y entrenamiento
lr_model = LinearRegression()
lr_model.fit(X_train_scaled, y_train)

# Predicciones
y_train_pred_lr = lr_model.predict(X_train_scaled)
y_test_pred_lr = lr_model.predict(X_test_scaled)

# Evaluación de métricas
mae_lr = mean_absolute_error(y_test, y_test_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test, y_test_pred_lr))
r2_lr = r2_score(y_test, y_test_pred_lr)

print(f"--- Regresión Lineal ---")
print(f"MAE:  {mae_lr:.4f}")
print(f"RMSE: {rmse_lr:.4f}")
print(f"R²:   {r2_lr:.4f}")
```

##Modelo 2: Árbol de Decisión

```python
from sklearn.tree import DecisionTreeRegressor

# Instanciación y entrenamiento con control de profundidad
dt_model = DecisionTreeRegressor(max_depth=10, random_state=42)
dt_model.fit(X_train_scaled, y_train)

# Predicciones
y_train_pred_dt = dt_model.predict(X_train_scaled)
y_test_pred_dt = dt_model.predict(X_test_scaled)

# Evaluación de métricas
mae_dt = mean_absolute_error(y_test, y_test_pred_dt)
rmse_dt = np.sqrt(mean_squared_error(y_test, y_test_pred_dt))
r2_dt = r2_score(y_test, y_test_pred_dt)

print(f"--- Árbol de Decisión ---")
print(f"MAE:  {mae_dt:.4f}")
print(f"RMSE: {rmse_dt:.4f}")
print(f"R²:   {r2_dt:.4f}")
```

##Modelo 3: Ensemble (Random Forest Regressor)

```python
from sklearn.ensemble import RandomForestRegressor

# Instanciación y entrenamiento del ensemble
rf_model = RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
rf_model.fit(X_train_scaled, y_train)

# Predicciones
y_train_pred_rf = rf_model.predict(X_train_scaled)
y_test_pred_rf = rf_model.predict(X_test_scaled)

# Evaluación de métricas
mae_rf = mean_absolute_error(y_test, y_test_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_test_pred_rf))
r2_rf = r2_score(y_test, y_test_pred_rf)

print(f"--- Random Forest ---")
print(f"MAE:  {mae_rf:.4f}")
print(f"RMSE: {rmse_rf:.4f}")
print(f"R²:   {r2_rf:.4f}")
```

# 9.- Consideraciones de Ética y Privacidad.

Usar Inteligencia Artificial en la música implica hacerlo con responsabilidad. Aunque el objetivo técnico es predecir qué tan popular puede ser una canción, debemos cuidar cómo se usan los datos y qué impacto tienen las predicciones en los oyentes y en los artistas.


##9.1. Privacidad y Protección de Datos


No se usan datos personales: El conjunto de datos solo contiene información técnica de las canciones (como duración, ritmo, volumen y género) y metadatos (como artista, álbum, nombre de la canción y género). No se utiliza ningún dato personal de los oyentes, como nombres, edades, ubicaciones o historiales de escucha.
Cumplimiento de las reglas de privacidad: Al trabajar exclusivamente con datos de las canciones y no de personas reales, se respeta la privacidad de los usuarios y se cumplen las leyes de protección de datos.


##9.2. Sesgos y Justicia en los Datos


Riesgo de favorecer solo a ciertos géneros: Si en los datos hay muchas más canciones de géneros comerciales (como Pop o Reggaeton), el modelo podría aprender que esa es la "única fórmula" para tener éxito, dejando en desventaja a géneros independientes o regionales.
Riesgo de perder la variedad musical: Si las discográficas o artistas crean música buscando únicamente darle el gusto al algoritmo (ajustando la duración o el ritmo a lo que el modelo pide), se corre el riesgo de que las canciones terminen sonando muy parecidas y se pierda la creatividad.


##9.3. Transparencia del Modelo


Saber cómo toma las decisiones: La Regresión Lineal es un modelo fácil de entender y explicar. A diferencia de otros modelos más complejos que funcionan como una "caja negra" (donde no se sabe bien qué pasa por dentro), aquí se puede ver claramente cuánto influye cada característica en el resultado final.
No mide la calidad artística: La predicción del modelo es solo un cálculo matemático basado en el audio. No sirve para decir si una canción es "buena" o "mala" desde el punto de vista del arte.


##9.4. Impacto en los Artistas y Uso Responsable


El éxito no depende solo del sonido: Que una canción sea popular en la vida real depende de muchas cosas que el modelo no puede ver, como el dinero invertido en publicidad, la fama del cantante o el impacto en redes sociales. Evaluar a un artista solo por lo que dice el algoritmo sería injusto.
Uso como guía, no como regla: Este modelo debe usarse solo como una herramienta de apoyo para entender tendencias, jamás como un filtro automático para decidir a qué artista contratar o apoyar.
